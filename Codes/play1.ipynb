### Exploratory Data Analysis

```python
num_cols = df_clean.select_dtypes(include=['int64', 'float64']).columns
```

```python
import seaborn as sns
```

```python
fig, axes = plt.subplots(nrows=4, ncols=3, figsize=(15, 12))
axes = axes.flatten()

for i, feature in enumerate(num_cols):
    sns.histplot(data=df_clean, x=feature, kde=True, ax=axes[i], color='coral')
    axes[i].set_title(f'Distribution of {feature}', fontsize=10)
    axes[i].set_xlabel("")

axes[-1].set_visible(False)

plt.tight_layout()
plt.show()
```

```python
corr_matrix = df_clean.corr()
mask = np.triu(np.ones_like(corr_matrix, dtype=bool))
plt.figure(figsize=(14, 10))

sns.heatmap(
    corr_matrix,
    mask=mask,
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    vmin=-1,
    vmax=1,
    square=True,
    linewidths=.5
)

plt.title('Multicollinearity Heatmap')
plt.show()
```

### Scaling and Model training

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(df_clean)
```

```python
pd.DataFrame(X_scaled, columns=df_clean.columns).describe().T
```

```python
from sklearn.cluster import KMeans

wcss = []

for k in range(1, 11):
    kmeans = KMeans(
        n_clusters=k,
        init='k-means++',
        random_state=42,
        n_init=10
    )

    kmeans.fit(X_scaled)
    wcss.append(kmeans.inertia_)
```

```python
plt.figure(figsize=(8,5))

plt.plot(range(1,11), wcss, marker='o')

plt.xlabel("Number of Clusters (K)")
plt.ylabel("WCSS")
plt.title("Elbow Method")

plt.show()
```

```python
from sklearn.metrics import silhouette_score

scores = []

for k in range(2,11):

    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = kmeans.fit_predict(X_scaled)

    score = silhouette_score(X_scaled, labels)

    scores.append(score)
```

```python
plt.figure(figsize=(8,5))

plt.plot(range(2,11), scores, marker='o')

plt.xlabel("Number of Clusters")
plt.ylabel("Silhouette Score")

plt.show()
```

```python
kmeans = KMeans(
    n_clusters=4,
    random_state=42,
    n_init=10
)

clusters = kmeans.fit_predict(X_scaled)
```

```python
df_clean['Cluster'] = clusters
```

```python
df_clean['Cluster'].value_counts()
```

```python
cluster_profile = df_clean.groupby('Cluster').mean()

cluster_profile
```

```python
plt.figure(figsize=(10,6))

sns.heatmap(cluster_profile,
            annot=True,
            cmap='coolwarm',
            fmt='.2f')

plt.title("Cluster Profiles")
plt.show()
```

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=2)

X_pca = pca.fit_transform(X_scaled)
```

```python
pca_df = pd.DataFrame({
    'PC1': X_pca[:,0],
    'PC2': X_pca[:,1],
    'Cluster': clusters
})
```

```python
import seaborn as sns

plt.figure(figsize=(8,6))

sns.scatterplot(
    data=pca_df,
    x='PC1',
    y='PC2',
    hue='Cluster',
    palette='Set2'
)

plt.title("Clusters Visualized Using PCA")
plt.show()
```

```python
from sklearn.metrics import (
    silhouette_score,
    davies_bouldin_score,
    calinski_harabasz_score
)

print("Silhouette Score :", silhouette_score(X_scaled, clusters))
print("Davies-Bouldin Index :", davies_bouldin_score(X_scaled, clusters))
print("Calinski-Harabasz Score :", calinski_harabasz_score(X_scaled, clusters))
```
